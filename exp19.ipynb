## Simple Port Scanner

A **port scanner** is a tool designed to probe a server or host for open ports. It can be used to determine what services are running on a machine and to identify potential vulnerabilities. This simple scanner will attempt to establish a connection to a specified range of ports on a target host. If a connection is successful, the port is considered open.

```python
import socket

def scan_port(host, port):
    """Attempts to connect to a given port on a host."""
    try:
        sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        sock.settimeout(1) # Set a timeout for the connection attempt
        result = sock.connect_ex((host, port))
        if result == 0:
            return True # Port is open
        else:
            return False # Port is closed or filtered
    except socket.error as e:
        # Handle connection errors (e.g., host unreachable)
        print(f"Error connecting to {host}:{port} - {e}")
        return False
    finally:
        sock.close()

def port_scanner(host, start_port, end_port):
    """Scans a range of ports on a target host."""
    print(f"Scanning ports {start_port}-{end_port} on {host}...")
    open_ports = []
    for port in range(start_port, end_port + 1):
        if scan_port(host, port):
            print(f"Port {port} is OPEN")
            open_ports.append(port)
        # else:
        #     print(f"Port {port} is CLOSED")

    if open_ports:
        print(f"\nScan completed. Open ports: {open_ports}")
    else:
        print("\nScan completed. No open ports found in the specified range.")
```

### Usage Example

To use the port scanner, you'll need to specify a target host (IP address or hostname) and a range of ports. For demonstration purposes, I will use `localhost` (127.0.0.1) and scan a small range of common ports. You might need to adjust the host and port range based on your network configuration and permissions.

**Note:** Scanning external IP addresses or large port ranges without explicit permission can be considered illegal and unethical. Only scan hosts you have permission to scan, such as your own local machine.

```python
# Example usage:
target_host = '127.0.0.1' # You can change this to a different IP or hostname
start_port = 20 # Starting port (e.g., FTP data port)
end_port = 80 # Ending port (e.g., HTTP port)

port_scanner(target_host, start_port, end_port)

# You can try scanning other common ports like:
# 21 (FTP control)
# 22 (SSH)
# 23 (Telnet)
# 25 (SMTP)
# 53 (DNS)
# 80 (HTTP)
# 110 (POP3)
# 143 (IMAP)
# 443 (HTTPS)
# 3389 (RDP)

# Example with a wider range for localhost (might take longer):
# port_scanner('127.0.0.1', 1, 1024)
```
