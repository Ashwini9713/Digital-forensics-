### Simple Machine Learning Model Example: Linear Regression

```python
# Import necessary libraries
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

First, let's create a simple synthetic dataset. This dataset will have a clear linear relationship between the input (X) and the output (y).

```python
# Generate synthetic data
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)

# Visualize the generated data
plt.scatter(X, y)
plt.xlabel('X (Input Feature)')
plt.ylabel('y (Target Variable)')
plt.title('Synthetic Dataset for Linear Regression')
plt.grid(True)
plt.show()
```

Now, we will train a `LinearRegression` model from `scikit-learn` on this dataset.

```python
# Create a Linear Regression model instance
model = LinearRegression()

# Train the model
model.fit(X, y)

print(f"Model Intercept: {model.intercept_[0]:.2f}")
print(f"Model Coefficient (slope): {model.coef_[0][0]:.2f}")
```

Finally, let's use the trained model to make a prediction for a new input value and visualize the regression line.

```python
# Make a prediction for a new value
X_new = np.array([[1.5]]) # Example new input
y_pred = model.predict(X_new)

print(f"Prediction for X = {X_new[0][0]}: {y_pred[0][0]:.2f}")

# Visualize the regression line
plt.scatter(X, y, label='Original Data')
plt.plot(X, model.predict(X), color='red', label='Regression Line')
plt.scatter(X_new, y_pred, color='green', marker='X', s=100, label='New Prediction')
plt.xlabel('X (Input Feature)')
plt.ylabel('y (Target Variable)')
plt.title('Linear Regression Model and Prediction')
plt.legend()
plt.grid(True)
plt.show()
```
