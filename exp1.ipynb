```python
import hashlib

def calculate_sha256(filepath, buffer_size=65536):
    """
    Calculates the SHA-256 hash of a file.

    Args:
        filepath (str): The path to the file.
        buffer_size (int): The size of the buffer to read the file in chunks.

    Returns:
        str: The SHA-256 hash of the file in hexadecimal format.
    """
    sha256_hash = hashlib.sha256()
    try:
        with open(filepath, "rb") as f:
            for byte_block in iter(lambda: f.read(buffer_size), b""):
                sha256_hash.update(byte_block)
        return sha256_hash.hexdigest()
    except FileNotFoundError:
        return "File not found."
    except Exception as e:
        return f"An error occurred: {e}"
```

### How to Use:

1.  **Replace `'your_file_path_here'`** with the actual path to the file you want to hash.
2.  **Run the cell** to calculate and print the SHA-256 hash.

This function reads the file in chunks, which is efficient for large files, preventing the entire file from being loaded into memory at once.

```python
# Example usage:
# Create a dummy file for demonstration
with open('sample_evidence.txt', 'w') as f:
    f.write('This is some sample data for hashing.\n')
    f.write('It will be used to demonstrate the SHA-256 function.')

file_to_hash = 'sample_evidence.txt'
hash_value = calculate_sha256(file_to_hash)

print(f"SHA-256 hash of '{file_to_hash}': {hash_value}")
```
